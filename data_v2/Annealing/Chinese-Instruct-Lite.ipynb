```python
import os
import glob
from datasets import load_dataset

def resolve_symlinks(file_pattern):
    """解析软链接并返回实际文件路径"""
    files = glob.glob(file_pattern)
    resolved_files = []
    
    for file_path in files:
        if os.path.islink(file_path):
            # 解析软链接
            link_target = os.readlink(file_path)
            if not os.path.isabs(link_target):
                # 相对路径，转换为绝对路径
                abs_target = os.path.join(os.path.dirname(file_path), link_target)
            else:
                abs_target = link_target
            
            if os.path.exists(abs_target):
                resolved_files.append(abs_target)
            else:
                print(f"警告: 软链接目标不存在: {abs_target}")
        else:
            resolved_files.append(file_path)
    
    return resolved_files

def get_all_subdirs(base_path):
    """获取基础路径下的所有子目录"""
    subdirs = []
    if os.path.exists(base_path):
        for item in os.listdir(base_path):
            item_path = os.path.join(base_path, item)
            if os.path.isdir(item_path):
                subdirs.append(item_path)
    return subdirs

def load_all_datasets(base_path):
    """从所有子目录中加载数据集文件"""
    print("=== 扫描所有子目录 ===")
    
    # 获取所有子目录
    subdirs = get_all_subdirs(base_path)
    print(f"找到的子目录数量: {len(subdirs)}")
    
    all_resolved_files = []
    datasets_info = {}
    
    # 遍历每个子目录
    for subdir in subdirs:
        subdir_name = os.path.basename(subdir)
        print(f"\n正在处理子目录: {subdir_name}")
        
        # 在当前子目录中查找parquet文件
        parquet_pattern = os.path.join(subdir, "*.parquet")
        resolved_files = resolve_symlinks(parquet_pattern)
        
        print(f"  - 找到的文件数量: {len(resolved_files)}")
        
        if resolved_files:
            # 记录每个子目录的信息
            datasets_info[subdir_name] = {
                'file_count': len(resolved_files),
                'files': resolved_files
            }
            # 添加到总文件列表
            all_resolved_files.extend(resolved_files)
            
            # 显示前几个文件名
            for i, file_path in enumerate(resolved_files[:3]):
                print(f"    文件 {i+1}: {os.path.basename(file_path)}")
            if len(resolved_files) > 3:
                print(f"    ... 还有 {len(resolved_files) - 3} 个文件")
    
    print(f"\n=== 汇总信息 ===")
    print(f"总共找到的文件数量: {len(all_resolved_files)}")
    
    # 显示每个子目录的统计信息
    for subdir_name, info in datasets_info.items():
        print(f"  - {subdir_name}: {info['file_count']} 个文件")
    
    return all_resolved_files, datasets_info

# 主要代码
base_path = "/DATA/disk2/yuhang/.cache/huggingface/hub/datasets--Mxode--Chinese-Instruct-Lite/snapshots/65ad0db337df2d03d63302e4e5811e0053508789"

# 获取所有文件
all_files, datasets_info = load_all_datasets(base_path)

if all_files:
    print(f"\n=== 开始加载数据集 ===")
    print(f"准备加载 {len(all_files)} 个文件...")
    
    try:
        # 加载所有数据集
        ds_complete = load_dataset(
            "parquet",
            data_files=all_files,
            split="train"
        )
        print(f"✅ 成功加载完整数据集！")
        print(f"总数据量: {len(ds_complete):,} 条记录")
        print(f"数据集列名: {list(ds_complete.column_names)}")
        
        # 显示第一条数据作为示例
        if len(ds_complete) > 0:
            print(f"\n=== 示例数据 ===")
            first_example = ds_complete[0]
            for key, value in first_example.items():
                if isinstance(value, str) and len(value) > 100:
                    print(f"{key}: {value[:100]}...")
                else:
                    print(f"{key}: {value}")
        
    except Exception as e:
        print(f"❌ 加载数据集时出错: {e}")
        print("\n建议：可以尝试分别加载各个子目录的数据")
        
        # 提供分别加载的选项
        print(f"\n=== 分别加载各子目录数据 ===")
        for subdir_name, info in datasets_info.items():
            try:
                print(f"正在加载 {subdir_name}...")
                ds_subset = load_dataset(
                    "parquet",
                    data_files=info['files'],
                    split="train"
                )
                print(f"  ✅ {subdir_name}: {len(ds_subset):,} 条记录")
            except Exception as subset_error:
                print(f"  ❌ {subdir_name}: 加载失败 - {subset_error}")
                
else:
    print("❌ 没有找到任何有效的parquet文件")
```

```python
# 筛选出 reasoning 字段为空的数据
print("正在筛选 reasoning 字段为空的数据...")

# 定义筛选函数：检查 reasoning 字段是否为空
def has_empty_reasoning(example):
    """
    检查 reasoning 字段是否为空
    
    Args:
        example: 数据集中的一条记录
        
    Returns:
        bool: 如果 reasoning 为空则返回 True，否则返回 False
    """
    reasoning = example['reasoning']
    # 检查是否为 None、空字符串或只包含空白字符
    return reasoning is None or reasoning.strip() == ""

# 使用 filter 方法筛选出 reasoning 为空的数据
ds_empty_reasoning = ds_complete.filter(has_empty_reasoning)

print(f"✅ 筛选完成！")
print(f"原始数据集大小: {len(ds_complete):,} 条")
print(f"reasoning 为空的数据: {len(ds_empty_reasoning):,} 条")
print(f"占比: {len(ds_empty_reasoning) / len(ds_complete) * 100:.2f}%")

# 显示几个示例
if len(ds_empty_reasoning) > 0:
    print(f"\n=== 前3个 reasoning 为空的数据示例 ===")
    for i in range(min(3, len(ds_empty_reasoning))):
        example = ds_empty_reasoning[i]
        print(f"\n示例 {i+1}:")
        print(f"ID: {example['id']}")
        print(f"prompt: {example['prompt'][:100]}...")
        print(f"reasoning: '{example['reasoning']}'")  # 显示空值
        print(f"response: {example['response'][:100]}...")
else:
    print("\n✅ 所有数据的 reasoning 字段都不为空！")
```

```python
# 从 ds_empty_reasoning 中随机采样 26% 的数据
print("正在从 reasoning 为空的数据中随机采样 26%...")

if len(ds_empty_reasoning) > 0:
    # 计算采样数量（26%）
    sample_size = int(len(ds_empty_reasoning) * 0.26)
    
    print(f"原始 reasoning 为空的数据量: {len(ds_empty_reasoning):,} 条")
    print(f"采样比例: 26%")
    print(f"采样数量: {sample_size:,} 条")
    
    # 使用 train_test_split 进行随机采样
    # 这里我们只需要采样部分，所以使用 train_test_split 的 test_size 参数
    from datasets import Dataset
    import random
    
    # 设置随机种子以确保结果可重现
    random.seed(42)
    
    # 随机采样 26% 的数据
    ds_sampled = ds_empty_reasoning.train_test_split(
        test_size=0.26,  # 采样 26%
        seed=42,         # 设置随机种子
        shuffle=True     # 确保随机打乱
    )['test']  # 取测试集作为我们的采样结果
    
    print(f"✅ 采样完成！")
    print(f"采样后数据量: {len(ds_sampled):,} 条")
    print(f"实际采样比例: {len(ds_sampled) / len(ds_empty_reasoning) * 100:.2f}%")
    
    # 显示采样数据的前几个示例
    print(f"\n=== 采样数据示例 ===")
    for i in range(min(2, len(ds_sampled))):
        example = ds_sampled[i]
        print(f"\n示例 {i+1}:")
        print(f"ID: {example['id']}")
        print(f"prompt: {example['prompt'][:80]}...")
        print(f"response: {example['response'][:80]}...")
    
    # 更新变量名，后续使用采样后的数据
    ds_empty_reasoning = ds_sampled
    print(f"\n📝 已将采样后的数据赋值给 ds_empty_reasoning 变量")
    
else:
    print("❌ 没有 reasoning 为空的数据，无法进行采样")
```

```python
# 统计 ds_empty_reasoning 中 'response' 字段的平均长度
print("正在统计 reasoning 为空数据的 response 字段长度...")

if len(ds_empty_reasoning) > 0:
    # 计算所有 response 字段的长度
    response_lengths = []
    
    for example in ds_empty_reasoning:
        response = example['response']
        # 确保 response 不为空，然后计算长度
        if response is not None:
            response_lengths.append(len(response))
        else:
            response_lengths.append(0)  # 如果 response 为空，长度记为 0
    
    # 计算统计信息
    total_responses = len(response_lengths)
    total_length = sum(response_lengths)
    average_length = total_length / total_responses if total_responses > 0 else 0
    min_length = min(response_lengths) if response_lengths else 0
    max_length = max(response_lengths) if response_lengths else 0
    
    print(f"\n=== reasoning 为空数据的 response 字段长度统计 ===")
    print(f"数据条数: {total_responses:,}")
    print(f"response 平均长度: {average_length:.2f} 字符")
    print(f"response 最短长度: {min_length:,} 字符")
    print(f"response 最长长度: {max_length:,} 字符")
    print(f"response 总字符数: {total_length:,} 字符")
    
    # 显示长度分布（可选）
    print(f"\n=== 长度分布示例 ===")
    print(f"前5个数据的 response 长度: {response_lengths[:5]}")
    
else:
    print("❌ 没有 reasoning 为空的数据，无法统计 response 长度")
```

```python
# 将 ds_empty_reasoning 中的 'prompt' 和 'response' 字段合并为 'text' 字段
print("正在合并 prompt 和 response 字段为 text 字段...")

if len(ds_empty_reasoning) > 0:
    def merge_prompt_response(example):
        """
        将 prompt 和 response 字段合并为 text 字段
        """
        # 获取 prompt 和 response 内容
        prompt = example['prompt'] if example['prompt'] is not None else ""
        response = example['response'] if example['response'] is not None else ""
        
        # 合并为 text 字段，使用换行符分隔
        text = f"{prompt}\n\n{response}"
        
        # 返回新的数据结构，包含 text 字段和其他保留字段
        return {
            'id': example['id'],
            'text': text,
            #'reasoning': example['reasoning']  # 保留 reasoning 字段
        }
    
    # 应用转换函数到数据集
    ds_merged = ds_empty_reasoning.map(merge_prompt_response,
                                       num_proc=30)
    
    print(f"✅ 成功合并字段！")
    print(f"原始字段: ['id', 'prompt', 'reasoning', 'response']")
    print(f"新字段: {list(ds_merged.features.keys())}")
    print(f"数据条数: {len(ds_merged):,}")
    
    # 显示合并后的示例数据
    print(f"\n=== 合并后的数据示例 ===")
    example = ds_merged[0]
    print(f"ID: {example['id']}")
    print(f"Text 前100字符: {example['text'][:100]}...")
    print(f"Text 总长度: {len(example['text']):,} 字符")
    print(f"Reasoning: '{example['reasoning']}'")
    
    # 更新变量名，方便后续使用
    ds_empty_reasoning = ds_merged
    print(f"\n✅ 数据集已更新为合并后的格式")
    
else:
    print("❌ 没有 reasoning 为空的数据，无需进行字段合并")
```

```python
# 将 ds_empty_reasoning 中的 'prompt' 和 'response' 字段合并为 'text' 字段
print("正在合并 prompt 和 reasoning 字段为 text 字段...")

if len(ds_reasoning) > 0:
    def merge_prompt_response(example):
        """
        将 prompt 和 response 字段合并为 text 字段
        """
        # 获取 prompt 和 response 内容
        prompt = example['prompt'] if example['prompt'] is not None else ""
        reasoning = example['reasoning'] if example['reasoning'] is not None else ""
        
        # 合并为 text 字段，使用换行符分隔
        text = f"{prompt}\n\n{reasoning}"
        
        # 返回新的数据结构，包含 text 字段和其他保留字段
        return {
            'id': example['id'],
            'text': text,
            #'reasoning': example['reasoning']  # 保留 reasoning 字段
        }
    
    # 应用转换函数到数据集
    ds_merged_reasoning = ds_reasoning.map(merge_prompt_response,
                                       num_proc=30)
    
    print(f"✅ 成功合并字段！")
    print(f"原始字段: ['id', 'prompt', 'reasoning', 'response']")
    print(f"新字段: {list(ds_merged_reasoning.features.keys())}")
    print(f"数据条数: {len(ds_merged_reasoning):,}")
    
    # 显示合并后的示例数据
    print(f"\n=== 合并后的数据示例 ===")
    example = ds_merged_reasoning[0]
    print(f"ID: {example['id']}")
    print(f"Text 前100字符: {example['text'][:100]}...")
    print(f"Text 总长度: {len(example['text']):,} 字符")
    print(f"Reasoning: '{example['reasoning']}'")
    
    print(f"\n✅ 数据集已更新为合并后的格式")
    
else:
    print("❌ 没有 reasoning 为空的数据，无需进行字段合并")
```
